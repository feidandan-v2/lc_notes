```python
class MyQueue:
    def __init__(self) -> None:
        self.stack_in = []
        self.stack_out = []
    
    def push(self, x: int) -> None:
        """new element push in"""
        self.stack_in.append(x)
    
    def pop(self,) -> int:
        """pop element out"""
        if self.stack_out:
            return self.stack_out.pop()
        else:
            for i in range(0, len(self.stack_in)):
                self.stack_out.append(self.stack_in[i])
            return self.stack_out.pop()
```
