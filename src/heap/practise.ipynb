```python
class MyQueue:
    def __init__(self) -> None:
        self.stack_in = []
        self.stack_out = []

    def push(self, x: int) -> None:
        return self.stack_in.append(x)
    
    def pop(self) -> int|None:
        if self.empty():
            return None
        if len(self.stack_out)!=0:
            return self.stack_out.pop()
        else:
            for i in range(0, len(self.stack_in)):
                self.stack_out.append(self.stack_in.pop())
                
            return self.stack_out.pop()
        
    def peek(self) -> int|None:
        res = self.pop()
        self.stack_out.append(res)

        return res

    def empty(self) -> bool:
        if len(self.stack_in) == 0 and  len(self.stack_out) == 0:
            return True
        else:
            return False
```

```python
queue = MyQueue()

queue.push(1)
queue.push(2)
queue.push(3)

print(queue.stack_in)
print(queue.stack_out)

print(queue.peek())

print(queue.stack_in)
print(queue.stack_out)
```

```python
if queue.stack_out:
    print(1, queue.stack_out)
else:
    print(2, queue.stack_out)

queue.pop()
```

```python
# 20.有效的括号

def isValid(s: str) -> bool:
    stack = []
    for item in s:
        if item == "(":
            stack.append(")")
        elif item == "[":
            stack.append("]")
        elif item == "{":
            stack.append("}")
        elif not stack:    # 栈提前pop完了，说明右括号多了
            return False, "right exceed"
        elif stack[-1] != item:  # 括号类型不匹配
            return False, "match problem"
        else:
            stack.pop()
    if not stack:         # 遍历完，栈正好清空
        return True
    else:
        return False, "left exceed"      # 遍历完 栈还有，说明左括号多了
```
