```python
class MyQueue:
    def __init__(self) -> None:
        self.stack_in = []
        self.stack_out = []

    def push(self, x: int) -> None:
        return self.stack_in.append(x)
    
    def pop(self) -> int|None:
        if self.empty():
            return None
        if len(self.stack_out)!=0:
            return self.stack_out.pop()
        else:
            for i in range(0, len(self.stack_in)):
                self.stack_out.append(self.stack_in.pop())
                
            return self.stack_out.pop()
        
    def peek(self) -> int|None:
        res = self.pop()
        self.stack_out.append(res)

        return res

    def empty(self) -> bool:
        if len(self.stack_in) == 0 and  len(self.stack_out) == 0:
            return True
        else:
            return False
```

```python
queue = MyQueue()

queue.push(1)
queue.push(2)
queue.push(3)

print(queue.stack_in)
print(queue.stack_out)

print(queue.peek())

print(queue.stack_in)
print(queue.stack_out)
```

```python
# 20.有效的括号

def isValid(s: str) -> bool:
    stack = []
    for item in s:
        if item == "(":
            stack.append(")")
        elif item == "[":
            stack.append("]")
        elif item == "{":
            stack.append("}")
        elif not stack:    # 栈提前pop完了，说明右括号多了
            return False, "right exceed"
        elif stack[-1] != item:  # 括号类型不匹配
            return False, "match problem"
        else:
            stack.pop()
    if not stack:         # 遍历完，栈正好清空
        return True
    else:
        return False, "left exceed"      # 遍历完 栈还有，说明左括号多了
```

```python
s = "([{}]()"
# s = "([{}})"
# s = "([{}]))"
# s = "[{}]()"
# s = "([)]"
res = isValid(s)
print(res)
```

```python
def remove_duplicates(s: str) -> str:
    stack = []
    for i in range(0, len(s)):
        if not stack:
            stack.append(s[i])
        else:
            if stack[-1] == s[i]:
                stack.pop()
            else:
                stack.append(s[i])
    return ''.join(stack)
```

```python
# 不用栈，尝试用双指针实现
def remove_duplicates1(s: str) -> str:
    left, right = 0, 0
    s = list(s)
    while right < len(s):
        s[left] = s[right]
        if left >0 and s[left] == s[left-1]:
            left -= 1
        else:
            left +=1
            
        right +=1

    return s[0:left]

s = "abbaac"
res = remove_duplicates1(s)
print(res)
```
